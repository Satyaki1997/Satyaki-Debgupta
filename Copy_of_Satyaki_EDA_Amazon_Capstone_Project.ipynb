**Project Summary: Exploratory Data Analysis for Amazon Prime TV Shows and Movies**

This project undertakes a detailed Exploratory Data Analysis (EDA) of Amazon Prime Video's content library, focusing on identifying trends, patterns, and key characteristics of its TV shows and movies. Utilizing two datasets – titles.csv and credits.csv – the analysis aims to provide data-driven insights to inform content strategy, acquisition, and audience engagement efforts.

The initial phase involved understanding the datasets through examining their structure, size, and variable types. The titles_df dataset provided metadata for nearly 10,000 titles, including genre, release year, runtime, and various popularity/rating scores (IMDb and TMDB). The credits_df dataset offered information about the cast and crew associated with each title. Initial checks revealed a significant number of missing values, particularly in age_certification and seasons columns within the titles_df, along with some duplicates across both datasets.

The Data Wrangling stage focused on cleaning and preparing the data for analysis. Recognizing the high percentage of missing data, the age_certification and seasons columns were dropped from the titles_df. Missing values in other columns, such as description, imdb_id, and character, were imputed with placeholder values like 'No description' or 'unknown' to maintain data integrity while acknowledging the missing information. Numerical missing values in rating and popularity scores (imdb_score, imdb_votes, tmdb_popularity, tmdb_score) were strategically filled using median imputation to mitigate the impact of outliers on the distribution. Duplicate rows in both the original and the merged datasets were identified and removed to ensure accurate counts and analyses. Finally, the titles_df and credits_df were merged on their common 'id' column, creating a comprehensive df DataFrame for integrated analysis, preserving all title information while linking it to associated cast and crew data.

The core of the project involved Data Visualization, Storytelling, and Experimentation with charts to understand the relationships between variables and uncover actionable insights. Univariate analysis revealed that the content library is heavily dominated by specific genres, with 'Drama' and 'Comedy' leading significantly in title count, indicating a concentration in these areas. The distribution of release_year showed a notable surge in content production in recent decades, peaking around 2020, reflecting the growth of streaming platforms. Analysis of imdb_score and tmdb_score distributions highlighted that most titles fall within an average rating range (around 5-7), with fewer titles achieving extremely high or low scores. Runtime analysis showed that the majority of content adheres to typical movie lengths, though with notable outliers.

Bivariate and multivariate analyses further enriched the understanding. Comparing content types revealed that while Movies constitute a larger volume, Shows exhibit higher average TMDB popularity, suggesting greater audience engagement per title for episodic content. Analyzing the correlation between numerical metrics indicated a moderate agreement between IMDb and TMDB scores, but weak correlations between these scores and popularity/vote counts, implying that critical reception, audience volume, and popularity are distinct dimensions of content performance. Exploring genre versus scores highlighted which genres tend to receive higher average ratings. Visualizing the distribution of content across production countries by type confirmed the dominance of regions like the US and India in content volume, while also showing how different countries contribute to the Movie and Show categories. Examining popularity by production country identified regions and, notably, joint country productions (like US-Japan or various European collaborations) that tend to yield content with higher average TMDB popularity.

The insights derived provide clear directions for achieving business objectives. To enhance library appeal and engagement, the platform should consider diversifying content acquisition beyond the most saturated genres, focusing on high-scoring titles regardless of genre, and strategically investing in Shows given their higher average popularity. Leveraging data on high-performing production countries and co-production models can refine acquisition targets. Tailoring marketing campaigns based on whether content is critically acclaimed, broadly popular, or highly engaging (based on vote counts) can improve reach and effectiveness. Furthermore, improving content discovery through recommendation algorithms that factor in these varied metrics can enhance user satisfaction and retention. Addressing imbalances in content types within popular genres or from specific regions could unlock new audience segments.

In conclusion, this EDA project provides a robust analytical foundation, moving beyond surface-level observations to reveal nuanced patterns in content distribution, audience reception, and production trends. The findings underscore the importance of a data-driven, balanced content strategy that optimizes acquisition, production, marketing, and discovery efforts to navigate the competitive streaming landscape effectively and drive business growth.

````python
#Missing Value Treatment

#Missing value treatment in titles_df

#missing values in age_certification are 65.7%..So we will drop it

titles_df = titles_df.drop('age_certification', axis=1)

````

````python

#missing values in seasons are 86.25%.so  this column should be dropped

titles_df= titles_df.drop('seasons', axis=1)
````

````python
# the missing values of tmdb_score will be replaced with mean values because moderate amount is missing

titles_df['tmdb_score'].fillna(titles_df['tmdb_score'].mean(), inplace=True)
````

````python
# missing values in description will be replaced with  'No description'

titles_df['description'].fillna('No description', inplace=True)
````

````python
# missing values of imdb_id will be replaced with 'unknown_imdb_id'

titles_df['imdb_id'].fillna('unknown_imdb_id', inplace=True)
````

````python
#missing values of imdb score will be replaced with median values
titles_df['imdb_score'].fillna(titles_df['imdb_score'].median(), inplace=True)
````

````python
#missing values of imdb votes will be replaced with median values

titles_df['imdb_votes'].fillna(titles_df['imdb_votes'].median(), inplace=True)
````

````python
#missing values of tmdb_popularity will be replaced with median values

titles_df['tmdb_popularity'].fillna(titles_df['tmdb_popularity'].median(), inplace=True)
````

````python
# in credits dataset missing values of characters will be replaced with 'unknown'

credits_df['character'].fillna('unknown', inplace=True)
````

````python
# Now before proceeding to final analysis I will merge titles_df and credits_df for obtaining the comprehensive clean dataset which will be used for plotting charts and drawing insights

df=pd.merge(titles_df,credits_df,on='id',how='left')

df.head()
````

````python
# Chart - 1 visualization code
# univariate analysis - Top 10 genres by count using bar chart analysis

# Calculate the top 10 genres by count
top10_genres = df['genres'].value_counts().head(10).reset_index()

# Rename the columns for clarity
top10_genres.columns = ['genre', 'count']

# Create the bar chart using Plotly Express
fig = px.bar(
    top10_genres,
    x='genre',  # Use the 'genre' column for the x-axis
    y='count',  # Use the 'count' column for the y-axis
    text='count',
    color='count',
    color_continuous_scale='Viridis',
    title='Top 10 Genres by Count on Amazon Prime Video',
)

# Beautify the layout
fig.update_traces(
    texttemplate='%{text}',
    textposition='outside',
    marker_line_color='black',
    marker_line_width=1.5
)
fig.update_layout(
    title_font_size=22,
    title_x=0.5,
    xaxis_title='Genre',
    yaxis_title='Number of Titles',
    xaxis=dict(tickangle=-30),
    plot_bgcolor='white',
    paper_bgcolor='white',
    font=dict(size=14),
    coloraxis_showscale=False
)

fig.show()
````

````python
#Heatmap of Content Count by Production Country & Type (Heatmap)

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Copy and prepare data
df_temp = df.copy()
df_temp['production_countries'] = df_temp['production_countries'].astype(str).str.split(',')
df_temp = df_temp.explode('production_countries')
df_temp['production_countries'] = df_temp['production_countries'].str.strip()

# Count by country & type
heat = df_temp.groupby(['production_countries', 'type']).size().reset_index(name='count')

# Pick top 10 countries
top_countries = (
    heat.groupby('production_countries')['count']
        .sum()
        .sort_values(ascending=False)
        .head(10)
        .index
)

heat_top = heat[heat['production_countries'].isin(top_countries)]

# Pivot table for heatmap
pivot = heat_top.pivot(index='production_countries', columns='type', values='count').fillna(0)

# Plot
plt.figure(figsize=(8,6))
sns.heatmap(pivot, annot=True, fmt='.0f', cmap='Blues')
plt.title("Content Count by Country & Type (Top 10 Countries)", fontsize=14)
plt.xlabel("Type")
plt.ylabel("Country")
plt.tight_layout()
plt.show()

````
